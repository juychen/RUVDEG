```python
import scanpy as sc
import os
```

```python
group = 'CUSUS_3v3_500_1000gene_new'
```

```python
adata = sc.read_h5ad(f'/data7/mark/STG/dataset/snRNA/merge_SCH_new/{group}/adata_SUS_F_3v3_downsampled_ratio.h5ad')
```

```python
ourdir = f'/data8/junyi/{group}/'
if not os.path.exists(ourdir):
    os.makedirs(ourdir)
```

```python
regions = adata.obs['region'].unique().tolist()
```

```python
for region in regions:
    adata_region = adata[adata.obs['region'] == region]
    adata_region.write_h5ad(f'{ourdir}/{region}.h5ad')
```
